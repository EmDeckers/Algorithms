Problem: Find the Nth node from the end of a Linked List

Approach: Two-Pointer Method
Use a fast and slow pointer. Set the fast pointer N steps ahead of the slow pointer, then simultaneously move them both forward. When the fast.next pointer is Null, the slow pointer will have stopped on the Nth node from the end of the list.
Check for edge cases along the way

```python
def nthNode(head: Node, N: int) -> Node:
    if not head:
        return None 
    if N < 1:
        return None
        
    current = head
    fast = head 

    for _ in range(N):
        if not fast:
            return
        else:
            fast = fast.next 

    while fast.next:
        current = current.next 
        fast = fast.next 

    return current.data
```

```python
def combineLists(list1: Node, list2:Node) -> Node:
    if list1 is None and list2:
        return list2
    elif list2 is None and list1:
        return list1
    elif list1 is None and list2 is None:
        return None

    head = None
    if list1.data <= list2.data:
        head = list1
        list1 = list1.next 
    else:
        head = list2
        list2 = list2.next 

    current_node = head 

    while list1 and list2:
        if list1.data <= list2.data:
            current_node.next = list1
            list1 = list1.next 
        else:
            current_node.next = list2
            list2 = list2.next 
            
        current_node = current_node.next 

    if list1:
        current_node.next = list1
    if list2:
        current_node.next = list2 

    return head 


def printList(head:Node):
    if head is None:
        print("There is no list")
        return
        
    node = head
    traversal = ""
    while node:
        traversal += (str(node.data) + "-> ")
        node = node.next

    print(traversal + "Null")
```

```python
class Solution:
    def reorderList(self, head: Optional[ListNode]) -> None:
        if not head or not head.next:
            return 

        slow = head 
        fast = head 

       # find the middle 
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next 

        prev, curr = None, slow.next 
        slow.next = None

        #reverse the list 
        while curr:
            nextNode = curr.next 
            curr.next = prev
            prev = curr
            curr = nextNode 

        # splice the lists together 
        first, second = head, prev
        while second:
            t1, t2 = first.next, second.next 
            first.next = second 
            second.next = t1
            first = t1
            second = t2
```

```python
class Solution:
    def isPalindrome(self, head: Optional[ListNode]) -> bool:
        if head is None:
            return False 
        
        if head.next is None:
            return True

        slow = head
        fast = head 

        # find the middle 
        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next 

        # reverse the second half 
        prev, current = None, slow

        while current:
            nextNode = current.next 
            current.next = prev 
            prev = current 
            current = nextNode 

        # compare node by node, stopping if they are not exactly identical
        while prev:
            if head.val != prev.val:
                return False 

            head = head.next 
            prev = prev.next 

        return True 
```
