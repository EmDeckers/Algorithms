Stacks 

```python
class Stack():
    def __init__(self):
        self.items = [] 

    def push(self, item):
        self.items.append(item)
        
    def pop(self):
        if not self.isEmpty():
            return self.items.pop()
            
    def peek(self):
        if not self.isEmpty():
            return self.items[-1]

    def getStack(self):
        return self.items

    def isEmpty(self):
        return self.items == []

    def __len__(self):
        if not self.isEmpty():
            return len(self.items)
```

```python
stack = Stack()
stack.push(1)
stack.push(2)
stack.push(4)
stack.push(6)
stack.push(7)

print(stack.getStack())

stack.pop()
stack.pop()

print(stack.getStack())
stack.isEmpty()
```

Write a function that determines whether a set of symbols has been properly matched and compiled 

Ex: s = "(({}))" returns True 
s = ")[](" returns False 

```python
def isValidSymbolString(s: str) -> bool:
    if not s:
        return False 

    opens = "([{"
    closes = "}])"
    stack = Stack()

    def getMatching(symbol: str) -> str:
        if not symbol:
            return

        if symbol == ")":
            return "("
        elif symbol == "]":
            return "["
        elif symbol == "}":
            return "{"

    for character in s:
        if character in opens:
            stack.push(character)
        elif character in closes and getMatching(character) == stack.peek():
            stack.pop()
        else:
            return False 

    return True 
```

```python
s = "(({}))" 
print(s, ":", isValidSymbolString(s))

s = ")[]("
print(s, ":", isValidSymbolString(s))
```

You are given a string of characters formed with a’s and b’s. 
The string is marked with special character X which represents the middle of the list.
Check whether the string is palindrome.

s ="abababaabababXbababbaaa" return False 

Strategy: find the middle, split, reverse the second half, then determine if first and second half are the same 

* Note: This could be solved using a two pointer method, but because the purpose of this notebook is to practice stacks, I will use a slightly more tedious stack implementation. 

```python
def isPalindrome(s:str) -> bool:
    if not s:
        return False 

    middle = s.index("X")
    first_half = s[:middle]
    second_half = s[middle+1:]

    if len(first_half) != len(second_half):
        return False

    stack = Stack()

    for character in second_half:
        stack.push(character)

    reverse = ""
    
    while not stack.isEmpty():
        char = stack.pop()
        reverse += char 

    if first_half == reverse:
        return True 
    else:
        return False 
```

```python
s = "ababXbababbaaa"

print(s, ": ", isPalindrome(s))

s = "abaabaaaXaaabaaba"
print(s, ": ", isPalindrome(s))
```

Implement a queue that supports enqueue, dequeue, and peek operations using only two stacks 

```python
class Queue():
    def __init__(self):
        self.s1 = Stack()
        self.s2 = Stack()

    def enQueue(self, item):
        if not self.s1.isEmpty():
            while not self.s1.isEmpty():
                self.s2.push(self.s1.pop())
    
            self.s2.push(item)
    
            while not self.s2.isEmpty():
                self.s1.push(self.s2.pop())
        else:
            self.s1.push(item)

        return self.s1.items

    def deQueue(self):
        self.s1.pop()
        return self.s1.items

    def peek(self):
        if not self.s1.isEmpty():
            return self.s1.peek()
```

```python
queue = Queue()
queue.enQueue(1)
queue.enQueue(4)
queue.enQueue(2)
queue.enQueue(7)
print("Q: ", queue.enQueue(5))

queue.deQueue()
print("Q: ", queue.deQueue())

print("Top element: ", queue.peek())
```

Write a function that converts a number to binary 

```python
import math 

def convertToBinary(number) -> int:
    number = math.floor(number)
    remainder = number 
    stack = Stack()
    
    while remainder > 0:
        if remainder % 2 == 0:
            stack.push(0)
        else:
            stack.push(1)

        remainder = remainder // 2

    binary = ""
    while not stack.isEmpty():
        binary += str(stack.pop())

    return int(binary)
```

```python
print("240.3 to binary: ", convertToBinary(240.3))
print("46 to binary: ", convertToBinary(46))
print("301.6 to binary: ", convertToBinary(301.6))
```
